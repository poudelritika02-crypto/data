### scatter plot

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Health_heart_experimental.csv")

# remove index column
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

plt.scatter(df["age"], df["BMI"])
plt.title("Scatter Plot of Age vs BMI")
plt.xlabel("Age")
plt.ylabel("BMI")
plt.show()



```

### histograph

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Health_heart_experimental.csv")

# remove index column if present
if "Unnamed: 0" in df.columns:
    df = df.drop(columns=["Unnamed: 0"])

plt.hist(df["age"], bins=10)
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

### box plot

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Health_heart_experimental.csv")

df = df.drop(columns=["Unnamed: 0"])

plt.boxplot(df.select_dtypes(include='number').values,
            labels=df.select_dtypes(include='number').columns)
plt.xticks(rotation=45)
plt.title("Box Plot of Heart Health Data")
plt.show()
```

### line chart

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Health_heart_experimental.csv")

# Count values of indication
counts = df['indication'].value_counts().sort_index()

# Create line chart
plt.figure()
plt.plot(counts.index, counts.values, marker='1')
plt.xlabel("Indication (0 = No Heart Disease, 1 = Heart Disease)")
plt.ylabel("Number of People")
plt.title("Line Chart of Heart Disease Indication")
plt.show()
```

### bar graph

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Health_heart_experimental.csv")

# Count values
counts = df['indication'].value_counts()

# Bar chart
plt.figure(figsize=(8, 6))
plt.bar(counts.index, counts.values, color='skyblue')
plt.xlabel('Indication')
plt.ylabel('Count')
plt.title('Heart Disease Indication Distribution')
plt.xticks(rotation=35)  # Rotate labels if necessary for readability
plt.tight_layout()  # Adjust layout to avoid label overlap
plt.show()
```

### pie chart

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Health_heart_experimental.csv")

# Count values of 'indication'
counts = df['indication'].value_counts()

# Pie chart with custom colors and explode feature
plt.figure(figsize=(7, 3))  # Making the chart a bit larger
explode = (0.1, 0)  # Exploding the first slice (Heart Disease)
colors = ['#ff9999', "#fffc66"]  # Custom colors for the slices

plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=colors, explode=explode)
plt.title("Heart Disease Indication Distribution")
plt.axis('equal')  # Ensures the pie chart is circular
plt.show()
```
